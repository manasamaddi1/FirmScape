# Week 6 – Exploratory Time-Series Analysis

**FirmScape:** How industry concentration and firm growth shape housing trajectories across U.S. **metros** (CBSA), and whether we can identify **leading indicators** of urban economic growth.

- **Unit of analysis:** Metro (CBSA), not city — e.g. SF + Oakland = Bay Area.
- **Quarter-based:** Aligns with earnings, hiring/layoffs, and migration.
- **Focus:** Are shifts in industry composition and concentration tied to housing price growth? Use **comparative cases** (Detroit, Bay Area, Austin, Seattle) to validate patterns.

**This notebook:**
1. Visualize industry/firm growth alongside housing price trends (metro and national).
2. Compare fast-growing metros to slower or declining ones (including Detroit vs tech hubs).
3. Explore correlations and **lagged** relationships (correlation matrix).
4. Examine differences across industry-concentration and diversity measures.

*Data: `firmscape_integrated_cbsa_quarterly.csv` (CBSA-quarter panel)*

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid", palette="muted")
%matplotlib inline

DATA_DIR = Path("..") / "data"
QFILE = DATA_DIR / "firmscape_integrated_cbsa_quarterly.csv"
```

```python
# Load quarterly panel (metro = CBSA, quarter)
df = pd.read_csv(QFILE)
df["yq"] = pd.to_datetime(df["yq"].astype(str).str.replace("Q", "-") + "-01", format="%Y-%m-%d", errors="coerce")
# Restrict to rows with housing + firm data for analysis
df_analysis = df.dropna(subset=["fhfa_index", "firms_founded_yoy"], how="any").copy()
df_analysis["year"] = df_analysis["year"].astype(int)

# Comparative case metros (rise/decline and tech hubs)
CASE_METROS = {
    "Detroit (MSAD)": 19804,           # Detroit-Dearborn-Livonia — manufacturing decline
    "San Francisco (MSAD)": 41884,     # Bay Area tech hub
    "Austin": 12420,                   # Austin-Round Rock-San Marcos — tech growth
    "Seattle (MSAD)": 42644,           # Seattle-Bellevue-Kent — tech hub
}

print("Shape:", df.shape)
print("Analysis subset (non-null fhfa + firms_founded_yoy):", df_analysis.shape)
df_analysis.head()
```

## 1. Industry growth alongside housing price trends

Plot aggregate (US) or selected metros: firm formation growth (`firms_founded_yoy`) and housing YoY (`fhfa_yoy`, `zillow_yoy`) over time.

```python
# National-level (average across CBSAs) quarterly series
agg = df_analysis.groupby("yq").agg(
    fhfa_yoy=("fhfa_yoy", "mean"),
    zillow_yoy=("zillow_yoy", "mean"),
    firms_founded_yoy=("firms_founded_yoy", "mean"),
    industry_count_new=("industry_count_new", "mean"),
).reset_index()

fig, axes = plt.subplots(2, 1, figsize=(12, 7), sharex=True)
ax1, ax2 = axes
ax1.plot(agg["yq"], agg["fhfa_yoy"] * 100, label="FHFA YoY %", color="C0")
ax1.plot(agg["yq"], agg["zillow_yoy"] * 100, label="Zillow YoY %", color="C1", alpha=0.8)
ax1.set_ylabel("Housing price growth (YoY %)")
ax1.legend(loc="upper right")
ax1.set_title("Housing price trends (avg across CBSAs)")
ax2.plot(agg["yq"], agg["firms_founded_yoy"], label="Firms founded YoY (ratio)", color="C2")
ax2.axhline(0, color="gray", linestyle="--", alpha=0.7)
ax2.set_ylabel("Firm formation growth")
ax2.set_xlabel("Quarter")
ax2.legend(loc="upper right")
ax2.set_title("Industry / firm formation growth (avg across CBSAs)")
plt.tight_layout()
plt.show()
```

```python
# Overlay: housing and firm growth on same scale (normalized for visibility)
agg_norm = agg.copy()
for c in ["fhfa_yoy", "zillow_yoy", "firms_founded_yoy"]:
    s = agg_norm[c].dropna()
    if s.abs().max() > 0:
        agg_norm[c + "_n"] = (s - s.mean()) / (s.std() + 1e-8)

fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(agg_norm["yq"], agg_norm["fhfa_yoy_n"], label="FHFA YoY (std)", alpha=0.9)
ax.plot(agg_norm["yq"], agg_norm["zillow_yoy_n"], label="Zillow YoY (std)", alpha=0.9)
ax.plot(agg_norm["yq"], agg_norm["firms_founded_yoy_n"], label="Firms founded YoY (std)", alpha=0.9)
ax.axhline(0, color="gray", linestyle="--", alpha=0.7)
ax.legend(loc="upper right")
ax.set_ylabel("Standardized value")
ax.set_title("Industry growth alongside housing (normalized)")
plt.tight_layout()
plt.show()
```

## 2. Compare fast-growing vs slower/declining metros

Compare **comparative cases**: Detroit (manufacturing decline) vs Bay Area, Austin, Seattle (tech hubs). Do firm growth and concentration patterns align with housing trajectories?

```python
# Subset to case metros
case_codes = list(CASE_METROS.values())
case_names = list(CASE_METROS.keys())
case_df = df_analysis[df_analysis["cbsa_code"].isin(case_codes)].copy()
case_df["metro_label"] = case_df["cbsa_code"].map(dict(zip(case_codes, case_names)))

fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)
# Housing YoY by metro
for label in case_names:
    sub = case_df[case_df["metro_label"] == label]
    axes[0].plot(sub["yq"], sub["fhfa_yoy"] * 100, label=label, alpha=0.9)
axes[0].axhline(0, color="gray", linestyle="--", alpha=0.7)
axes[0].set_ylabel("FHFA housing YoY (%)")
axes[0].legend(loc="upper right")
axes[0].set_title("Housing price growth: Detroit vs tech hubs (metro-level)")
# Firm formation YoY by metro
for label in case_names:
    sub = case_df[case_df["metro_label"] == label]
    axes[1].plot(sub["yq"], sub["firms_founded_yoy"], label=label, alpha=0.9)
axes[1].axhline(0, color="gray", linestyle="--", alpha=0.7)
axes[1].set_ylabel("Firms founded YoY (ratio)")
axes[1].set_xlabel("Quarter")
axes[1].legend(loc="upper right")
axes[1].set_title("Firm formation growth: Detroit vs tech hubs")
plt.tight_layout()
plt.show()
```

## 3. Correlations and lagged relationships

Correlation matrix: housing growth (FHFA YoY, Zillow YoY) vs firm growth and **concentration** (HHI, industry diversity). Include lags to look for **leading indicators** (e.g. firm growth at t−k vs housing at t).

```python
# Variables for correlation (housing, firm growth, concentration/diversity, lags)
corr_cols = [
    "fhfa_yoy", "zillow_yoy",
    "firms_founded_yoy", "firms_founded_yoy_lag1q", "firms_founded_yoy_lag4q", "firms_founded_yoy_lag8q",
    "hhi_new", "hhi_new_lag1q", "industry_count_new",
    "fhfa_yoy_fwd4q", "zillow_yoy_fwd4q",
]
available = [c for c in corr_cols if c in df_analysis.columns]
corr_df = df_analysis[available].dropna(how="all")

# Short labels for plot
labels = {
    "fhfa_yoy": "FHFA YoY",
    "zillow_yoy": "Zillow YoY",
    "firms_founded_yoy": "Firm growth",
    "firms_founded_yoy_lag1q": "Firm growth lag 1q",
    "firms_founded_yoy_lag4q": "Firm growth lag 4q",
    "firms_founded_yoy_lag8q": "Firm growth lag 8q",
    "hhi_new": "HHI (conc.)",
    "hhi_new_lag1q": "HHI lag 1q",
    "industry_count_new": "Industry count",
    "fhfa_yoy_fwd4q": "FHFA YoY +4q",
    "zillow_yoy_fwd4q": "Zillow YoY +4q",
}
corr = corr_df[available].corr()
corr = corr.rename(index=labels, columns=labels)

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt=".2f", center=0, cmap="RdBu_r", ax=ax, vmin=-0.3, vmax=0.3)
ax.set_title("Correlation matrix: housing, firm growth, concentration, and lags (metro-quarter)")
plt.tight_layout()
plt.show()
```

```python
# Leading indicator check: does firm growth (or HHI) at t-k predict housing at t?
# Correlation of lagged firm growth with current and forward housing
lead_summary = []
for lag in ["firms_founded_yoy", "firms_founded_yoy_lag1q", "firms_founded_yoy_lag4q", "firms_founded_yoy_lag8q"]:
    if lag not in df_analysis.columns:
        continue
    for target in ["fhfa_yoy", "fhfa_yoy_fwd4q"]:
        if target not in df_analysis.columns:
            continue
        r = df_analysis[[lag, target]].corr().iloc[0, 1]
        lead_summary.append({"predictor": lag, "target": target, "correlation": r})
pd.DataFrame(lead_summary)
```

## 4. Differences across industry concentration and diversity

Metros with **higher industry diversity** (lower HHI, higher industry_count) or **higher concentration** (higher HHI) may show different housing–growth relationships. Compare distribution of housing growth by concentration/diversity buckets.

```python
# Diversity: 1 - top_industry_share_new (higher = more diverse); HHI (higher = more concentrated)
plot_df = df_analysis.dropna(subset=["fhfa_yoy", "hhi_new", "industry_count_new"]).copy()
plot_df["diversity_proxy"] = 1 - plot_df["top_industry_share_new"]
plot_df["hhi_tercile"] = pd.qcut(plot_df["hhi_new"], 3, labels=["Low conc.", "Mid conc.", "High conc."])
plot_df["industry_count_tercile"] = pd.qcut(plot_df["industry_count_new"].clip(upper=100), 3, labels=["Low div.", "Mid div.", "High div."])

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
sns.boxplot(data=plot_df, x="hhi_tercile", y="fhfa_yoy", ax=axes[0])
axes[0].axhline(0, color="gray", linestyle="--", alpha=0.7)
axes[0].set_ylabel("FHFA YoY")
axes[0].set_title("Housing growth by industry concentration (HHI tercile)")
sns.boxplot(data=plot_df, x="industry_count_tercile", y="fhfa_yoy", ax=axes[1])
axes[1].axhline(0, color="gray", linestyle="--", alpha=0.7)
axes[1].set_ylabel("FHFA YoY")
axes[1].set_title("Housing growth by industry diversity (industry count tercile)")
plt.tight_layout()
plt.show()
```

## Key patterns (interpretation)

- **Industry growth vs housing:** Do firm formation and housing move together in time? Do lagged firm (or entry_emp) growth correlate with future housing (leading indicator)?
- **Comparative cases:** Detroit vs Austin/Seattle/SF — do tech hubs show stronger firm–housing co-movement or different concentration patterns?
- **Concentration/diversity:** Are high-HHI (concentrated) metros associated with higher or lower housing growth variance? Does diversity (industry count) correlate with housing growth?
- **Tradeoffs:** Quarterly data and metro aggregation smooth city-level noise; missing firm vars in early years can reduce sample for lag analysis.
