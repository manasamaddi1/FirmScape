# **FirmScape**

FirmScape is a multi-scape data analysis and visualization project that examines how industry concentration and company growth shape the rise and decline of U.S. cities over time. By analyzing historial company and housing data, we aim to understand the dynamics of urban development and identify key factors influencing city growth and decline. 

In this notebook, we will be showcasing our company and housing data, the diversity and scoring of the correlation between the cities and housing prices, and our statistical tests and takeaways from our analysis.

# 1. Data Sources & Merges

For FirmScape, there are 4 different datasets (2 company, 2 housing) that were used to create our final dataset. The datasets are as follows:

1) "7+ Million Company Dataset" People Data Labs 2019 Global Company Dataset  
        --> Name, Industry, Size Range, Locality are important columns to focus on within this dataset
2) "17MIL Global Companies" SavvyIQ BigPicture 
        --> name, industry, size, state are important columns to focus on within this dataset 
3) "FHFA House Price Index" 
        --> city, state, cbsa_code, hpi_index, pct_change are important columns to focus on within this dataset
4) "ZHVI Data" Zillow  
        --> RegionName, StateName, date, price are important columns to focus on within this dataset

The final merged dataset combines the company and housing data, allowing us to analyze the relationship between industry concentration and housing market trends.

```
# "7+ Million Company Dataset" Cleaning

# Read CSV
df = pd.read_csv("companies_sorted.csv")

# Filter to US companies with 100+ employees
us_companies = df[
    (df["country"].str.strip().str.lower() == "united states") &
    (df["current employee estimate"] >= 100)
]

# Drop unwanted columns
us_companies = us_companies.drop(columns=[
    "Unnamed: 0",
    "linkedin url",
    "current employee estimate",
    "total employee estimate",
    "country"
    "domain"
], errors="ignore")

# Save to new CSV
us_companies.to_csv("us_companies_100plus.csv", index=False)

print("Saved us_companies_100plus.csv")
print("Final shape:", us_companies.shape)
# Preview
print(us_companies.head())

```

```
#"17MIL Global Companies" Cleaning

import pandas as pd

# Read CSV
df = pd.read_csv("companies-2023-q4-sm.csv", dtype=str)

# Normalize fields used for filtering
df["country_code"] = df["country_code"].str.strip().str.upper()
df["size"] = df["size"].str.strip()
df["founded"] = df["founded"].str.strip()


# US companies only
df = df[df["country_code"] == "US"]

# 100+ employees
size_100_plus = [
    "51-200",
    "201-500",
    "1K-5K",
    "5K-10K",
    "10K+"
]
df = df[df["size"].isin(size_100_plus)]

# Founded year exists
df = df[df["founded"].str.contains(r"\d{4}", regex=True, na=False)]

# Keep required columns
df = df[["handle", "name", "industry", "size", "state"]]

print("Final rows, columns:", df.shape)
print(df.head())

# Filtered CSV
df.to_csv("companies_us_100plus_clean.csv", index=False)

```

```
#"FHFA House Price Index" Cleaning

# 1. Load the dataset
column_names = ['location', 'cbsa_code', 'year', 'quarter', 'hpi_index', 'pct_change']
df = pd.read_csv('hpi_at_metro.csv', names=column_names)

def clean_housing_data(df):
    """
    Cleans FHFA metro-level HPI data and returns a sorted time-series
    with recomputed percent change per city/state.
    """

    df = df.copy()

    # Convert numeric fields
    df['hpi_index'] = pd.to_numeric(df['hpi_index'].replace('-', np.nan), errors='coerce')

    # Split location into city/state
    geo_split = df['location'].str.split(',', n=1, expand=True)
    df['city'] = geo_split[0].str.strip().str.lower()
    df['state'] = geo_split[1].str.strip().str.upper()

    # Create datetime column
    quarter_to_month = {1: '01', 2: '04', 3: '07', 4: '10'}
    df['date'] = pd.to_datetime(
        df['year'].astype(str) + '-' + df['quarter'].map(quarter_to_month) + '-01',
        errors='coerce'
    )

    # Drop rows without HPI and sort BEFORE pct_change
    df = (
        df
        .dropna(subset=['hpi_index', 'date'])
        .sort_values(['city', 'state', 'date'])
        .copy()
    )

    # Recompute percent change
    df['pct_change'] = (
        df
        .groupby(['city', 'state'])['hpi_index']
        .pct_change() * 100
    )

    final_cols = ['date', 'city', 'state', 'cbsa_code', 'hpi_index', 'pct_change']
    return df[final_cols].reset_index(drop=True)

# Apply cleaning
cleaned_df = clean_housing_data(df)

# Preview
print(cleaned_df.head())
```

```
#"ZHVI Data" Cleaning

# 1. Load the dataset
file_path = 'Zillow_Housing_Dataset.csv'
df = pd.read_csv(file_path)

# 2. View the first 5 rows to see the columns
print("--- First 5 Rows ---")
display(df.head())

# 3. Check data types and missing values
print("\n--- Dataset Info ---")
df.info()

# 4. List all column names (helpful for the next step of cleaning)
print("\n--- Column Names ---")
print(df.columns.tolist())
```

```
id_vars = ['RegionID', 'SizeRank', 'RegionName', 'RegionType', 'StateName']
zillow_long = pd.melt(df, id_vars=id_vars, var_name='date', value_name='price')

# Convert the date column to actual dates
zillow_long['date'] = pd.to_datetime(zillow_long['date'])

print("Zillow data is now aligned!")
display(zillow_long)
```

```
# 1. Sort the data so time flows correctly for each city
zillow_long = zillow_long.sort_values(by=['RegionName', 'date'])

# 2. Fill small gaps (Forward Fill)- carries the last known price forward
zillow_long['price'] = zillow_long.groupby('RegionName')['price'].ffill()

# 3. Drop the remaining NaNs
zillow_final = zillow_long.dropna(subset=['price'])

print(f"Removed {zillow_long['price'].isna().sum()} empty rows.")
print("--- Final Cleaned Data ---")
display(zillow_final.head())

# Save the cleaned Zillow data to a CSV file
zillow_final.to_csv('cleaned_zillow_data.csv', index=False)

print("File 'cleaned_zillow_data.csv' has been exported!")
```

```
#Final Cleaned Dataset (EDA)
import pandas as pd

# Load CSV files
companies = pd.read_csv("companies_us_100plus_clean.csv")
housing = pd.read_csv("cleaned_housing_data.csv")

# Dictionary to convert state names to abbreviations
state_map = {
    "Alabama": "AL", "Alaska": "AK", "Arizona": "AZ", "Arkansas": "AR",
    "California": "CA", "Colorado": "CO", "Connecticut": "CT", "Delaware": "DE",
    "Florida": "FL", "Georgia": "GA", "Hawaii": "HI", "Idaho": "ID",
    "Illinois": "IL", "Indiana": "IN", "Iowa": "IA", "Kansas": "KS",
    "Kentucky": "KY", "Louisiana": "LA", "Maine": "ME", "Maryland": "MD",
    "Massachusetts": "MA", "Michigan": "MI", "Minnesota": "MN",
    "Mississippi": "MS", "Missouri": "MO", "Montana": "MT", "Nebraska": "NE",
    "Nevada": "NV", "New Hampshire": "NH", "New Jersey": "NJ",
    "New Mexico": "NM", "New York": "NY", "North Carolina": "NC",
    "North Dakota": "ND", "Ohio": "OH", "Oklahoma": "OK", "Oregon": "OR",
    "Pennsylvania": "PA", "Rhode Island": "RI", "South Carolina": "SC",
    "South Dakota": "SD", "Tennessee": "TN", "Texas": "TX", "Utah": "UT",
    "Vermont": "VT", "Virginia": "VA", "Washington": "WA",
    "West Virginia": "WV", "Wisconsin": "WI", "Wyoming": "WY",
    "District of Columbia": "DC"
}

# Standardize company states
companies["state"] = companies["state"].map(state_map)

# Convert housing date
housing["date"] = pd.to_datetime(housing["date"])

# Latest housing row per state (keeps pct_change, hpi_index, cbsa_code, etc.)
latest_housing = housing.loc[
    housing.groupby("state")["date"].idxmax()
]

# Date range per state
date_range = (
    housing.groupby("state")["date"]
           .agg(["min", "max"])
           .reset_index()
)

date_range.columns = ["state", "earliest_housing_date", "latest_housing_date"]

# Merge latest housing + date range
housing_final = pd.merge(
    latest_housing,
    date_range,
    on="state",
    how="left"
)

# Merge with companies
merged = pd.merge(
    companies,
    housing_final,
    on="state",
    how="left"
)

# Save
merged.to_csv("merged_companies_housing.csv", index=False)
```

```
#Final Cleaned Dataset (Ranked List)
#Dictionary with states mapped to abbreviation
STATE_TO_ABBR = {
    "alabama": "AL", "alaska": "AK", "arizona": "AZ", "arkansas": "AR",
    "california": "CA", "colorado": "CO", "connecticut": "CT",
    "delaware": "DE", "district of columbia": "DC", "florida": "FL",
    "georgia": "GA", "hawaii": "HI", "idaho": "ID", "illinois": "IL",
    "indiana": "IN", "iowa": "IA", "kansas": "KS", "kentucky": "KY",
    "louisiana": "LA", "maine": "ME", "maryland": "MD", "massachusetts": "MA",
    "michigan": "MI", "minnesota": "MN", "mississippi": "MS", "missouri": "MO",
    "montana": "MT", "nebraska": "NE", "nevada": "NV", "new hampshire": "NH",
    "new jersey": "NJ", "new mexico": "NM", "new york": "NY", "north carolina": "NC",
    "north dakota": "ND", "ohio": "OH", "oklahoma": "OK", "oregon": "OR",
    "pennsylvania": "PA", "rhode island": "RI", "south carolina": "SC",
    "south dakota": "SD", "tennessee": "TN", "texas": "TX", "utah": "UT",
    "vermont": "VT", "virginia": "VA", "washington": "WA", "west virginia": "WV",
    "wisconsin": "WI", "wyoming": "WY",
}
# Helper functions
def _size_to_mid(s: str) -> float:
    """Map size range to midpoint (for median_size_mid)."""
    if pd.isna(s) or not isinstance(s, str):
        return np.nan
    s = s.strip().upper()
    if "10K+" in s or "10001+" in s:
        return 15000.0
    if "1K-5K" in s or "1001-5000" in s:
        return 3000.0
    if "201-500" in s or "501-1000" in s:
        return 500.0
    if "51-200" in s:
        return 125.5
    if "11-50" in s:
        return 30.0
    if "1-10" in s:
        return 5.0
    return np.nan

def _is_large_size(s: str) -> bool:
    """True if size is 1K+ (for share_large_proxy)."""
    if pd.isna(s) or not isinstance(s, str):
        return False
    s = s.strip().upper()
    return "1K" in s or "10K" in s or "10001+" in s or "1001" in s
# Load hpi data
hpi_next = pd.read_csv('hpi_at_metro.csv', header=None, names=["metro_name", "cbsa_code", "year", "quarter", "hpi_value", "hpi_se"]).copy()
hpi_next["fhfa_index"] = pd.to_numeric(hpi_next["hpi_value"].replace("-", np.nan).replace("", np.nan), errors="coerce")
hpi_next["cbsa_code"] = pd.to_numeric(hpi_next["cbsa_code"], errors="coerce")
hpi_next["year"] = pd.to_numeric(hpi_next["year"], errors="coerce").astype("Int64")
hpi_next["quarter"] = pd.to_numeric(hpi_next["quarter"], errors="coerce").astype("Int64")
hpi_next = hpi_next.drop(columns=["hpi_value", "hpi_se"], errors="ignore")
hpi_next = hpi_next.dropna(subset=["cbsa_code", "year", "quarter"])

# Add growth rates
hpi_next = hpi_next.sort_values(["cbsa_code", "year", "quarter"])
hpi_next["fhfa_qoq"] = hpi_next.groupby("cbsa_code")["fhfa_index"].pct_change(1, fill_method=None)
hpi_next["fhfa_yoy"] = hpi_next.groupby("cbsa_code")["fhfa_index"].pct_change(4, fill_method=None)
metro_cbsa = hpi_next[["metro_name", "cbsa_code"]].drop_duplicates()
print(hpi_next.head(5))

# Load zillow data
zillow_next = pd.read_csv('cleaned_zillow_data.csv')
zillow_next["date"] = pd.to_datetime(zillow_next["date"], errors="coerce")
zillow_next = zillow_next.dropna(subset=["date", "price"])
zillow_next["year"] = zillow_next["date"].dt.year
zillow_next["quarter"] = zillow_next["date"].dt.quarter
zillow_q = zillow_next.groupby(["RegionName", "year", "quarter"], as_index=False)["price"].mean()
zillow_q = zillow_q.rename(columns={"price": "zillow_price_q"})
print(zillow_q.head(5))

# Load company dataset
companies_next = pd.read_csv("companies_us_100plus_clean.csv")
companies_next = companies_next[companies_next['country_code'].astype(str).str.upper().eq("US")].copy()
def to_abbr(s):
    s = str(s).strip()
    if len(s) == 2:
        return s.upper()
    return STATE_TO_ABBR.get(s.lower(), "")
companies_next["state_abbr"] = companies_next["state"].astype(str).apply(to_abbr)
companies_next["city_norm"] = companies_next["city"].astype(str).str.strip()
companies_next["metro_key"] = companies_next["city_norm"] + ", " + companies_next["state_abbr"]
mc = metro_cbsa[["metro_name", "cbsa_code"]].drop_duplicates()
companies_next = companies_next.merge(mc, left_on="metro_key", right_on="metro_name", how="left")
companies_next = companies_next.drop(columns=["metro_key", "state_abbr", "city_norm", "metro_name"], errors="ignore")
companies_next["founded"] = pd.to_numeric(companies_next["founded"], errors="coerce")
companies_next = companies_next.dropna(subset=["cbsa_code"])
print(companies_next.head(5))
```

```
# Compute stock metrics (overall metrics) for all metros
c = companies_next.copy()
c["size_mid"] = c["size"].astype(str).apply(_size_to_mid)
c["is_large"] = c["size"].astype(str).apply(_is_large_size)
stock_df = c.groupby("cbsa_code").agg(
    firm_count_total=("name", "count"),
    industry_list=("industry", list),
    share_large_proxy=("is_large", "mean"),
    median_size_mid=("size_mid", "median"),
).reset_index()

def hhi_top(ind_list):
    flat = [i for i in ind_list if isinstance(i, str)]
    if not flat:
        return np.nan, np.nan, 0
    vc = pd.Series(flat).value_counts()
    n = len(vc)
    shares = vc / vc.sum()
    hhi = (shares ** 2).sum()
    top = shares.iloc[0]
    return hhi, top, n

stock_df["hhi_stock"] = stock_df["industry_list"].apply(lambda x: hhi_top(x)[0])
stock_df["top_industry_share_stock"] = stock_df["industry_list"].apply(lambda x: hhi_top(x)[1])
stock_df["industry_count_stock"] = stock_df["industry_list"].apply(lambda x: hhi_top(x)[2])
stock_df = stock_df.drop(columns=["industry_list"])
stock_df.head()
```

```
# Agrregate firms by CBSA-quarter to find quarterly statistics
companies_flow = companies_next.copy()
companies_flow["founded"] = pd.to_numeric(companies_flow["founded"], errors="coerce")
companies_flow = companies_flow.dropna(subset=["founded"])
companies_flow["year"] = companies_flow["founded"].astype(int)
companies_flow["quarter"] = 1  # Assign all to Q1

firms_agg = companies_flow.groupby(["cbsa_code", "year", "quarter"]).agg(
    firms_founded=("name", "count"),
    industry_list=("industry", list),
).reset_index()

def hhi_top(ind_list):
    flat = [i for i in ind_list if isinstance(i, str)]
    if not flat:
        return np.nan, np.nan, 0
    vc = pd.Series(flat).value_counts()
    n = len(vc)
    shares = vc / vc.sum()
    hhi = (shares ** 2).sum()
    top = shares.iloc[0]
    return hhi, top, n

firms_agg["hhi_new"] = firms_agg["industry_list"].apply(lambda x: hhi_top(x)[0])
firms_agg["top_industry_share_new"] = firms_agg["industry_list"].apply(lambda x: hhi_top(x)[1])
firms_agg["industry_count_new"] = firms_agg["industry_list"].apply(lambda x: hhi_top(x)[2])
firms_agg = firms_agg.drop(columns=["industry_list"])
firms_agg["firms_founded_yoy"] = firms_agg.groupby("cbsa_code")["firms_founded"].pct_change(4, fill_method=None)
firms_agg.head()
```

```
# Merge datasets
skel = hpi_next[["cbsa_code", "metro_name", "year", "quarter", "fhfa_index", "fhfa_qoq", "fhfa_yoy"]].copy()
skel["city_state"] = skel["metro_name"]

# Merge Zillow (match RegionName to metro_name)
zq = zillow_q.merge(metro_cbsa[["metro_name", "cbsa_code"]].drop_duplicates(), 
                    left_on="RegionName", right_on="metro_name", how="inner")
zq = zq.groupby(["cbsa_code", "year", "quarter"], as_index=False)["zillow_price_q"].mean()
zq["zillow_qoq"] = zq.groupby("cbsa_code")["zillow_price_q"].pct_change(1, fill_method=None)
zq["zillow_yoy"] = zq.groupby("cbsa_code")["zillow_price_q"].pct_change(4, fill_method=None)
skel = skel.merge(zq, on=["cbsa_code", "year", "quarter"], how="left")

# Merge firms flow metrics
skel = skel.merge(firms_agg, on=["cbsa_code", "year", "quarter"], how="left")

# Merge stock metrics (constant per metro)
skel = skel.merge(stock_df, on="cbsa_code", how="left")

skel["yq"] = skel["year"].astype(str) + "Q" + skel["quarter"].astype(str)
print(f"Merged skeleton: {len(skel)} rows")
skel.head()
```

```
# Add lagged values
skel = skel.sort_values(["cbsa_code", "year", "quarter"])

for lag in [1, 2, 4, 8]:
    skel[f"firms_founded_yoy_lag{lag}q"] = skel.groupby("cbsa_code")["firms_founded_yoy"].shift(lag)
    skel[f"hhi_new_lag{lag}q"] = skel.groupby("cbsa_code")["hhi_new"].shift(lag)


skel["fhfa_yoy_fwd4q"] = skel.groupby("cbsa_code")["fhfa_yoy"].shift(-4)
skel["zillow_yoy_fwd4q"] = skel.groupby("cbsa_code")["zillow_yoy"].shift(-4)

# Ensure stock metrics are present
if "share_large_proxy" not in skel.columns:
    skel["share_large_proxy"] = np.nan
if "median_size_mid" not in out.columns:
    skel["median_size_mid"] = np.nan
skel.head()

```

### What's in the final merged table?

The merged table (`skel`) is a **CBSA × quarter panel** — one row per metro per quarter. Here are the key fields:

#### Geography & Time
- **`cbsa_code`**: Metro identifier (Core Based Statistical Area code)
- **`metro_name`**: Metro name in "City, ST" format
- **`city_state`**: Alias for metro_name
- **`year`**, **`quarter`**: Time index (we'll add `yq` and `date` columns in the cleaning step)

#### Housing Signals
- **`fhfa_index`**: FHFA Housing Price Index level
- **`fhfa_qoq`**: Quarter-over-quarter growth rate (FHFA)
- **`fhfa_yoy`**: Year-over-year growth rate (FHFA)
- **`zillow_price_q`**: Quarterly mean Zillow price level (in dollars)
- **`zillow_qoq`**: Quarter-over-quarter growth rate (Zillow)
- **`zillow_yoy`**: Year-over-year growth rate (Zillow)

#### Firm Entry (Flow Metrics)
- **`firms_founded`**: Count of firms founded in that metro × year (all assigned to Q1)
- **`firms_founded_yoy`**: Year-over-year growth rate in new firm formation
- **`hhi_new`**: Herfindahl-Hirschman Index (industry concentration) among **new firms** in that quarter
- **`top_industry_share_new`**: Share of the largest industry among new firms
- **`industry_count_new`**: Number of distinct industries among new firms

#### Metro Stock Structure (Constant per Metro)
- **`firm_count_total`**: Total count of firms mapped to that metro (across all time)
- **`hhi_stock`**: HHI across **all firms** in the metro (not just new ones)
- **`top_industry_share_stock`**: Largest industry share across all firms
- **`industry_count_stock`**: Number of distinct industries across all firms
- **`share_large_proxy`**: Fraction of firms classified as "large" (1K+ employees)
- **`median_size_mid`**: Median firm size midpoint in the metro

---

**What this structure enables:**
- ✅ Time-series analysis: track how housing, firm formation, and industry mix evolve over quarters
- ✅ Lagged relationships: test whether firm entry leads housing growth (or vice versa)
- ✅ Metro-level scoring: aggregate to one row per metro for composite rankings
- ✅ Comparative analysis: compare metros like Detroit, Bay Area, Austin, Seattle over time

## Firm Size Distribution

Because our dataset states company size through size buckets, we convert these into numeric estimates to test these relationships statistically

```
def convert_size(value):
    if isinstance(value, str):
        value = value.replace(',', '').strip()
        
        if '+' in value:
            value = value.replace('+', '')
            if 'K' in value:
                return float(value.replace('K', '')) * 1000
            else:
                return float(value)
        
        if '-' in value:
            low, high = value.split('-')
            
            def parse_number(x):
                x = x.strip()
                if 'K' in x:
                    return float(x.replace('K', '')) * 1000
                else:
                    return float(x)
            
            return (parse_number(low) + parse_number(high)) / 2
    
    return None
```

```
df['size_numeric'] = df['size'].apply(convert_size)
print(f"\nOriginal size values (sample):\n{df['size'].value_counts().head(10)}")
print(f"\nConverted size statistics:")
print(df['size_numeric'].describe())
```

## Industry Concentration & Diversity Across Cities

To measure how economically diversified each city is, we use the Herfindahl-Hirschman Index (HHI).
The HHI is calculated by summing the squared market share of each industry within a city.

- HHI close to 1 → One industry dominates the city
- HHI close to 0 → The city is highly diversified across industries

Investigating a city's economic diversity is important because diversified cities are more resilient innovation hubs.
If one industry declines (e.g., tech layoffs, oil price crashes, manufacturing downturns), diversified cities are better positioned to absorb economic shocks, sustain employment, and maintain growth.

Industry diversity is therefore a core component of long-term economic strength and startup ecosystem durability.


```
df = pd.read_csv("merged_companies_housing.csv")

if 'industry' in df.columns:
    
    # Compute company counts per city
    city_totals = df.groupby('city_x').size().reset_index(name='total')
    
    # Filter: Keep only cities with at least N companies
    MIN_COMPANIES = 10   # You can change this threshold
    valid_cities = city_totals[city_totals['total'] >= MIN_COMPANIES]['city_x']
    
    df_filtered = df[df['city_x'].isin(valid_cities)].copy()
    
    print(f"Total cities before filtering: {city_totals.shape[0]}")
    print(f"Cities with >= {MIN_COMPANIES} companies: {len(valid_cities)}")
    
    # Calculate HHI

    city_industry = (
        df_filtered
        .groupby(['city_x', 'industry'])
        .size()
        .reset_index(name='count')
    )
    
    city_totals_filtered = (
        df_filtered
        .groupby('city_x')
        .size()
        .reset_index(name='total')
    )
    
    city_industry = city_industry.merge(city_totals_filtered, on='city_x')
    
    city_industry['share'] = city_industry['count'] / city_industry['total']
    city_industry['share_squared'] = city_industry['share'] ** 2
    
    hhi = (
        city_industry
        .groupby('city_x')['share_squared']
        .sum()
        .reset_index()
    )
    
    # Visualize
    hhi.columns = ['city', 'hhi']
    hhi = hhi.sort_values('hhi', ascending=False)
    
    print(f"\nTop 10 most concentrated cities (by HHI):")
    print(hhi.head(10))
    
    print(f"\nTop 10 most diversified cities (by HHI):")
    print(hhi.tail(10))
    
    print(f"\nMedian HHI (filtered cities): {hhi['hhi'].median():.3f}")
    
    fig, ax = plt.subplots(figsize=(14, 7))
    
    ax.hist(hhi['hhi'], bins=50, color='salmon',
            edgecolor='black', alpha=0.7)
    
    ax.set_xlabel('Herfindahl-Hirschman Index (HHI)', fontsize=11)
    ax.set_ylabel('Number of Cities', fontsize=11)
    ax.set_title(
        f'Distribution of Industry Concentration Across Cities\n',
        fontsize=13, fontweight='bold'
    )
    
    median_hhi = hhi['hhi'].median()
    ax.axvline(median_hhi,
               color='red',
               linestyle='--',
               linewidth=2,
               label=f'Median = {median_hhi:.3f}')
    
    ax.legend()
    plt.tight_layout()
    plt.show()
```

The median HHI across filtered cities is:

**Median HHI = 0.066**

This indicates that the typical city in our dataset is relatively diversified. An HHI of 0.066 suggests that no single industry dominates economic activity, and most cities have firms distributed across multiple sectors.

The distribution appears right-skewed, with most cities clustered at low HHI values and a smaller number of cities exhibiting moderate concentration.

The cities with the highest industry concentration are:

- Scott – 0.540
- Lake Park – 0.472
- Broussard – 0.429

Even among the most concentrated cities, HHI values remain below 0.6, indicating that while one industry may have a dominant share, some level of diversification still exists.

The relatively high value for New York City likely reflects dataset composition, where one industry category (e.g., finance or technology) may represent a disproportionately large share of recorded firms.

The most diversified cities are:

- Cleveland – 0.0161
- Los Angeles – 0.0166
- Casablanca – 0.0168

These cities demonstrate extremely balanced industry representation, with no single sector accounting for a large share of firms. Such low HHI values indicate strong economic diversification.

The majority of urban economies in the dataset appear diversified rather than specialized. While some cities exhibit moderate industry concentration, broad multi-sector representation is more common.

This supports the use of HHI as a meaningful component in our broader city scoring framework, as it captures structural economic balance rather than simply firm counts.

# 4. Composite Next-Hub + Housing Opportunity Score 

To determine the Next-Hub + Housing Opportunity Score, we define 4 key metrics:

- **Company Growth**
> We look at the rate at which new firms are formed in metros across the United States. This is done by looking at the year-over-year growth rate in new firm formation in the **'firms_founded_yoy'** column of the dataset. We take the **mean** of the year-over-year rates **(2010+)**  as our company growth variable.

- **Industry Diversity** 
> We look at the number of distinct industries amongst new firms founded in its particular metro. This is found in the **'industry_count_new'** column and we take the **mean** of this column for quarters from **2010+** as our industry diversity variable.

- **Housing Price Growth**
> We look at the year-over-year growth rate in housing prices across each metro for quarters from 2010. We accomplish this by using the **'fhfa_yoy'** value as our main measurement and the **'zillow_yoy'** as a **fallback** for rows missing the 'fhfa_yoy' value. We take the **mean** of these measurements from quarters **2010+** as our housing price growth variable.

- **Housing Affordability**
> We look at the affordability of a metro by using negative prices as it allows for cheaper houses to have a higher score. We use **-('zillow_price_q')** value as our main measurement and the **-('fhfa_index')** as a **fallback** for rows missing the 'zillow_price_q value'. We take the **mean** of these values for each metro from **2010+** as our housing affordability variable.

# Normalized Data
To create a score we normalize our data to keep each variable between [0, 1].
We do this through min-max scaling for the following reasons:
- **Bounds** each value between **[0, 1]** to align with the range of our score 
- This allows for 0 to be the worst and 1 to be the best
- The min-max **prevents** from variables with **larger values** having **higher weight** than other variables. The yoy rates, counts, and prices have **very different scales** so normalizing the data allows for these variables to be **properly scaled**.
4. Normalized data **does not assume** a **normal distribution curve** for our dataset, which it is not due to a multitude of **confounding variables**

# Weights
We use weigh each of the four variables **equally** and add them together to determine the composite score. We do this by multiplying each normalized variable by 1/4, meaning that no variable is preferred over others.
Why?
- We use equal weights as a **transparent, assumption-light** baseline because there is no validated ground truth for next-hub outcomes; unequal weights would introduce **unvalidated** value judgments, whereas equal weights provide **interpretability** and a clean baseline for sensitivity analysis

# Interpretation 
All metros have a composite score between **[0, 1]** as a result of normalizing each variable and weighing it equally. A higher composite score suggests that metro is a stronger candidate as the next hub.
A metro with a high composite score:
- Has **strong** firm formation growth
- Has a **diverse** industry mix (not just a monopoly, less concentration risk)
- Shows housing price **momentum** 
- Remains relatively affordable (**accessible** to new workers/residents)

Tradeoffs:
- High growth metros may have lower affordability
- Affordable metros may have slower growth


# Components and Normalization

```
q_data = pd.read_csv("firmscape_integrated_quarterly_cleaned.csv").copy()

if "date" in q_data.columns:
    q_data["date"] = pd.to_datetime(q_data["date"])
elif "yq" in q_data.columns:
    q_data["date"] = pd.to_datetime(
        q_data["yq"].astype(str).str.replace("Q", "-") + "-01",
        format="%Y-%m-%d",
        errors="coerce",
    )

q_data = q_data[q_data["year"] >= 2010].copy()

agg_cols = {
    "firms_founded_yoy": "mean",
    "industry_count_new": "mean",
    "top_industry_share_new": "mean",
    "fhfa_yoy": "mean",
    "zillow_yoy": "mean",
    "fhfa_index": "mean",
    "zillow_price_q": "mean",
}

use = {k: v for k, v in agg_cols.items() if k in q_data.columns}

next_hub = q_data.groupby("cbsa_code").agg(use).reset_index()

names = q_data.groupby("cbsa_code")["metro_name"].first()
next_hub["metro_name"] = next_hub["cbsa_code"].map(names)

next_hub = next_hub.dropna(subset=list(use.keys()), how="all")
next_hub.head(10)
```

```
import pandas at pd
def normalize_0_1(series: pd.Series) -> pd.Series:
    """Min-max to [0, 1]. Handles NaN."""
    s = series.dropna()
    if s.empty or s.min() == s.max():
        return series
    return (series - s.min()) / (s.max() - s.min())

# Raw component columns
next_hub["company_growth"] = next_hub["firms_founded_yoy"]
next_hub["industry_diversity"] = next_hub["industry_count_new"].fillna(0)
next_hub["housing_price_growth"] = next_hub["fhfa_yoy"].fillna(next_hub["zillow_yoy"])
# Affordability: higher = better → use negative price level (lower price = higher score)
next_hub["housing_affordability"] = -next_hub["zillow_price_q"]
next_hub["housing_affordability"] = next_hub["housing_affordability"].fillna(-next_hub["fhfa_index"])

# Normalized [0, 1] (higher = better for all after definitions above)
for col in ["company_growth", "industry_diversity", "housing_price_growth", "housing_affordability"]:
    next_hub[col + "_n"] = normalize_0_1(next_hub[col])

# Composite: equal weight (or weight by preference)
w1, w2, w3, w4 = 0.25, 0.25, 0.25, 0.25
next_hub["composite_score"] = (
    w1 * next_hub["company_growth_n"] +
    w2 * next_hub["industry_diversity_n"] +
    w3 * next_hub["housing_price_growth_n"] +
    w4 * next_hub["housing_affordability_n"]
)
next_hub = next_hub.dropna(subset=["composite_score"])
print("Normalized components and composite (sample):")
next_hub[["metro_name", "company_growth_n", "industry_diversity_n", "housing_price_growth_n", "housing_affordability_n", "composite_score"]].head(20)
```

### Sanity checks

- All normalized components (`*_n`) are in **[0, 1]**.
- No NaN inflation: where a raw component was non-missing, the normalized version is also non-missing.
- `composite_score` is a weighted average of [0, 1] components, so it also lies in [0, 1].
- Metros with completely missing components are dropped from the composite ranking (we don’t assign scores based on pure NaNs).

This ranking answers: **Which metros best balance firm growth, industry diversity, housing price momentum, and housing affordability under our equal-weighted definition?**

Metros are sorted by `composite_score` (descending), so rank 1 is the strongest “next hub + housing opportunity” candidate in this framework.

## Alternative ranking: where firm formation and housing growth move together

This is a **different question** from “best opportunity.” Here we ask:
> In which metros are **firm formation growth** and **housing price growth** most tightly synchronized over time?

We rank metros by the **Pearson correlation** between quarterly `firms_founded_yoy` and `fhfa_yoy` (2010+), within each metro.

### Variables Used in the Correlation-Based Ranking

For the correlation-based ranking, we focus on **how strongly firm formation growth and housing price growth move together over time within each metro**. This uses the **quarterly panel (2010+)**, not metro-level means.

We rely on two main time-series variables:

1. Firm Formation Growth (`firms_founded_yoy`)
    We measure how rapidly new firm formation is changing over time within a metro.
    
    - For each CBSA × quarter, `firms_founded` counts how many firms were founded (all assigned to Q1 of their founded year).
    - Within each metro, we compute the **year-over-year growth rate** in new firm counts:
      - `firms_founded_yoy = groupby(cbsa_code)["firms_founded"].pct_change(4, fill_method=None)`
    - This gives a quarterly time series of firm entry growth, starting from 2010+.
    
    In the correlation ranking, `firms_founded_yoy` is treated as the **firm-side signal**: higher values mean faster growth in new firms in that quarter.

2. Housing Price Growth (`fhfa_yoy`)
    We measure how housing prices are changing over time within the same metro.
    
    - From the FHFA HPI, we compute the **year-over-year growth rate** in the housing price index:
      - `fhfa_yoy = groupby(cbsa_code)["fhfa_index"].pct_change(4, fill_method=None)`
    - This gives a quarterly time series of housing price growth (post-2010) for each metro.
    - If `fhfa_yoy` is missing for some rows, those rows are excluded from the correlation calculation; we do **not** substitute Zillow here to keep the series conceptually consistent.

3. Correlation Variable (`firm_housing_corr`)
    To capture **synchronization** between firm growth and housing growth within each metro, we compute a Pearson correlation:

    - We first restrict to quarters where **both** `firms_founded_yoy` and `fhfa_yoy` are non-null.
    - For each `cbsa_code`, we compute:
      - `firm_housing_corr = corr(firms_founded_yoy, fhfa_yoy)`  
      - **Only if** there are at least 8 quarters with both values present; otherwise we set the correlation to `NaN`.
    - This yields one correlation coefficient per metro, in the range [-1, 1]:
      - +1 = perfectly aligned (firm growth and housing growth move together)
      - 0  = no linear relationship
      - -1 = perfectly inverse relationship

The **correlation-based ranking** then:

- Sorts metros by `firm_housing_corr` (descending), with highest correlation at the top.
- Assigns `rank = 1, 2, …` based on this ordering.
- Treats metros with `NaN` correlations (insufficient data) as valid rows but ranks them at the bottom.

In words, this ranking answers:  
> “Where do **firm formation growth** and **housing price growth** move together most strongly over time, based on their within-metro correlation?”

```
q_data_corr = q_data.dropna(subset=["firms_founded_yoy", "fhfa_yoy"]).copy()

corr_by_metro = q_data_corr.groupby("cbsa_code", group_keys=False).apply(
    lambda g: g["firms_founded_yoy"].corr(g["fhfa_yoy"]) if len(g) >= 8 else np.nan,
    include_groups=False,
).reset_index()
corr_by_metro.columns = ["cbsa_code", "firm_housing_corr"]

metro_with_corr = next_hub.merge(corr_by_metro, on="cbsa_code", how="left")

metro_with_corr["corr_rank"] = metro_with_corr["firm_housing_corr"].rank(
    ascending=False,
    method="min",
    na_option="bottom",
)

rank_by_corr = (
    metro_with_corr.sort_values("corr_rank")
    [["metro_name", "firm_housing_corr", "industry_diversity_n", "composite_score"]]
    .reset_index(drop=True)
)
rank_by_corr.insert(0, "rank", range(1, len(rank_by_corr) + 1))

print(f"Correlation-based ranking: {len(rank_by_corr)} metros")
print(f"Metros with valid correlation (≥8 quarters): {rank_by_corr['firm_housing_corr'].notna().sum()}")
print(f"Metros with NaN correlation (insufficient data): {rank_by_corr['firm_housing_corr'].isna().sum()}")
rank_by_corr.head(20)
```

### How to interpret this correlation ranking

- **High positive correlation** (near +1) suggests that firm entry and housing price growth tend to move together over time in that metro — one plausible story is that new firm activity brings workers and demand pressure into the housing market.
- **Low or near-zero correlation** suggests that firm dynamics and housing growth are decoupled or driven by different factors (e.g., macro shocks, supply constraints).
- **Negative correlation** would mean that when firm entry is strong, housing growth is weak (or vice versa), which is interesting from a policy / structural perspective.

 **Important:** Correlation is only computed for metros with **at least 8 quarters** where both `firms_founded_yoy` and `fhfa_yoy` are present. Many smaller metros don’t meet this threshold and show `NaN` in the table; they are pushed to the bottom of the ranking rather than silently dropped.

# 5. Hypothesis Testing 

We use an independent samples t-test to compare company growth between metros with high vs. low industry diversity.

**Group Split**: Metros are divided at the median of industry_diversity (industry_count_new).

Group A: (High Diversity): industry_diversity >= median
Group B: (Low Diversity): industry_diversity < median

Hypotheses: 
H0: Mean company growth is the same in both groups (mu_A = mu_B)
H1: Mean company growth differs between groups (mu_A != mu_B) 
We run an equal variance test first, which determines whether to use the standard or 2 sample t-test.

Significance Level: Alpha = 0.05 

```
from scipy import stats
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np

ALPHA = 0.05

# Prepare Data
ht_df = metro.dropna(subset=["industry_diversity", "company_growth"]).copy()
print(f"Metros available: {len(ht_df)}")

# Split at median industry diversity
median_id = ht_df["industry_diversity"].median()
high_div = ht_df.loc[ht_df["industry_diversity"] >= median_id, "company_growth"]
low_div  = ht_df.loc[ht_df["industry_diversity"] <  median_id, "company_growth"]

print(f"\nMedian industry diversity split: {median_id:.4f}")
print(f"  High-diversity group (n={len(high_div)}): mean company growth = {high_div.mean()*100:.3f}%")
print(f"  Low-diversity  group (n={len(low_div)}):  mean company growth = {low_div.mean()*100:.3f}%")

# test for equal variances 
lev_stat, lev_p = stats.levene(high_div, low_div)
equal_var = lev_p >= ALPHA
print(f"\nLevene's test: stat={lev_stat:.4f}, p={lev_p:.4f} -> "
      f"{'Equal variances assumed (standard t-test)' if equal_var else 'Unequal variances -> Welch t-test'}")

# Independent samples t-test
t_stat, t_p = stats.ttest_ind(high_div, low_div, equal_var=equal_var)
reject = t_p < ALPHA

# Effect size: Cohen's d
pooled_std = np.sqrt((high_div.std()**2 + low_div.std()**2) / 2)
cohens_d = (high_div.mean() - low_div.mean()) / pooled_std
effect_label = "large" if abs(cohens_d) > 0.8 else ("medium" if abs(cohens_d) > 0.5 else "small")

# Results
print("\n" + "="*60)
print("  INDEPENDENT SAMPLES T-TEST RESULTS")
print("="*60)
print(f"  t-statistic : {t_stat:+.4f}")
print(f"  p-value     : {t_p:.4f}")
print(f"  Cohen's d   : {cohens_d:+.4f}  ({effect_label} effect)")
print(f"  Decision    : {'REJECT H0 - significant difference' if reject else 'FAIL TO REJECT H0 - no significant difference'}")
print("="*60)

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Boxplot
ax = axes[0]
bp = ax.boxplot(
    [high_div.values, low_div.values],
    labels=["High Industry\nDiversity", "Low Industry\nDiversity"],
    patch_artist=True,
    notch=True,
    medianprops=dict(color="black", linewidth=2)
)
bp["boxes"][0].set_facecolor("mediumseagreen")
bp["boxes"][0].set_alpha(0.6)
bp["boxes"][1].set_facecolor("tomato")
bp["boxes"][1].set_alpha(0.6)
ax.yaxis.set_major_formatter(mticker.PercentFormatter(xmax=1, decimals=1))
ax.set_ylabel("Company Growth (Firms Founded YoY)", fontsize=10)
ax.set_title("Company Growth by\nIndustry Diversity Group", fontsize=11)
ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.annotate(
    f"t={t_stat:+.3f}, p={t_p:.3f}\n{'*Significant*' if reject else 'Not significant'}",
    xy=(0.5, 0.95), xycoords="axes fraction", ha="center", va="top",
    fontsize=9, color="darkred" if reject else "gray",
    bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="gray", alpha=0.8)
)

# KDE distribution
ax2 = axes[1]
for grp, label, color in [
    (high_div, "High Industry Diversity", "mediumseagreen"),
    (low_div,  "Low Industry Diversity",  "tomato")
]:
    kde = stats.gaussian_kde(grp)
    x_range = np.linspace(grp.min(), grp.max(), 300)
    ax2.plot(x_range, kde(x_range), label=f"{label} (n={len(grp)})", color=color, linewidth=2)
    ax2.axvline(grp.mean(), color=color, linestyle="--", linewidth=1.2, alpha=0.7)
    ax2.fill_between(x_range, kde(x_range), alpha=0.15, color=color)
ax2.xaxis.set_major_formatter(mticker.PercentFormatter(xmax=1, decimals=1))
ax2.set_xlabel("Company Growth (Firms Founded YoY)", fontsize=10)
ax2.set_ylabel("Density", fontsize=10)
ax2.set_title("Distribution of Company Growth\nby Industry Diversity Group", fontsize=11)
ax2.legend(fontsize=9)
ax2.grid(linestyle="--", alpha=0.4)

plt.suptitle(
    f"T-Test: Company Growth — High vs Low Industry Diversity Metros\n"
    f"Cohen's d = {cohens_d:+.3f} ({effect_label} effect)",
    fontsize=12, y=1.02
)
plt.tight_layout()
plt.show()

# Interpretation
print("\nInterpretation:")
if reject:
    direction = "higher" if t_stat > 0 else "lower"
    print(f"  We REJECT H0 (t={t_stat:+.3f}, p={t_p:.4f} < {ALPHA}).")
    print(f"  Metros with HIGH industry diversity show significantly {direction} company growth.")
    print(f"  Effect size (Cohen's d={cohens_d:+.3f}) is {effect_label}.")
else:
    print(f"  We FAIL TO REJECT H0 (t={t_stat:+.3f}, p={t_p:.4f} >= {ALPHA}).")
    print(f"  There is NO statistically significant difference in company growth")
    print(f"  between high and low industry diversity metros at alpha={ALPHA}.")
    print(f"  Effect size (Cohen's d={cohens_d:+.3f}) is {effect_label}.")
```
